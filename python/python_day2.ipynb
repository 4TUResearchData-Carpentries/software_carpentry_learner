```python
## Analysis and visualization of tabular data with numpy 
import numpy as np 
### Import the data ###

data=np.loadtxt(fname="data/inflammation-01.csv",delimiter=",")

print(data.shape)

#### analyze data #####

maxval,minval,stdval,meanval=np.amax(data,axis=0),np.amin(data,axis=0),np.std(data,axis=1),np.mean(data,axis=1)
print("max inflammation:",maxval)
print("min inflammation:",len(minval))
print("standard inflammation:",len(stdval))
print("average inflammation:", len(meanval))
```

```python
### Visualizing data ####

import matplotlib.pyplot as plt 

plt.plot(np.mean(data,axis=0))## average of all patients per day 
plt.plot(np.mean(data,axis=1)) ## average of all days per patient 
```

```python
## Analyse data from multiple files 

import glob 

## use glob to find all inflmmation data files 
file_pattern="data/inflammation-*.csv"
all_files= glob.glob(file_pattern)

##Initialize an empty list to store data arrays 

data_arrays=[]

## Loop through all matching files and load them as numpy arrays 

for file in all_files:
    print(file)
    data=np.loadtxt(file,delimiter=",")
    data_arrays.append(data)

### Analysis and plotting 

for datasets in data_arrays:
    mean_perday=np.mean(datasets,axis=0)
    max_perday=np.amax(datasets,axis=0)
    min_perday=np.amin(datasets,axis=0)
    ## Create the figure of three subplots 
    fig=plt.figure(figsize=(15.0,3.0))
    ## Plot in subplots
    axes1=fig.add_subplot(1,3,1)
    axes2=fig.add_subplot(1,3,2)
    axes3=fig.add_subplot(1,3,3)

    axes1.plot(mean_perday)
    axes2.plot(max_perday)
    axes3.plot(min_perday)

    axes1.set_ylabel("mean")
    axes2.set_ylabel("Max")
    axes3.set_ylabel("Min")
     
```
