## Lists and operations

We can create a `list` by putting values inside square brackets and separating values with commas

```python
odds = [1, 3, 5, 7]
print('odds in list are: ', odds)
```

We can access **items** in the **list** using **0-based index**

```python
print('first element: ', odds[0])
print('last element:', odds[3])
```
