## Lists and operations

We can create a `list` by putting values inside square brackets and separating values with commas

```python
odds = [1, 3, 5, 7]
print('odds in list are: ', odds)
```

We can access **items** in the **list** using **0-based index**

```python
print('first element: ', odds[0])
print('last element:', odds[3])
```

We can also do negative indexes to access items starting from the end of the list **(negative indexes are not 0-based)**

```python
print('last element: ', odds[-1])
print('second to last element:', odds[-2])
```

List items can also be modified using their indexes as values

```python
# typo on Darwin's name
names = ['Curie', 'Darwing', 'Turing']
print("names are: ", names)
```

```python
names[1] = 'Darwin'
print("correct names are", names)
```

Strings can also be accessed as lists, but their characters cannot be changed as items in a list

```python
name = 'Darwin'
name[0] = 'd'
```

Data which can be modified in place is called **mutable**, while data which cannot be modified is called **immutable**.

**Strings and numbers are immutable**.

**Lists are mutable**.

Lists can caontain any python variable, icluding other lists. These are called **nested lists**
