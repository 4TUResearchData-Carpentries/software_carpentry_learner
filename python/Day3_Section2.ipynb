# Creating functions

So far we've using **built-in** functions (`print()`) and **library functions** (`numpy.loadtxt()`). Here we'll lear to do our own

Let's calculate temperature

```python
fahrenheit_val = 99
celsius_val = ((fahrenheit_val -32) * (5/9))
celsius_val
```

For getting more conversions, we can copy paste and modify the input

```python
fahrenheit_val = 43
celsius_val = ((fahrenheit_val -32) * (5/9))
celsius_val
```

Functions are a better way to reuse functionality

```python
def fahr_to_celsius(temp):
    return((temp - 32)*(5/9))
```

```python
print(fahr_to_celsius(43))
print(fahr_to_celsius(99))
```

We can also write Celsius into Kelvin

```python
def celsius_to_kelvin(temp_c):
    return temp_c + 273.15

celsius_to_kelvin(27)
```

What about Fahrenheit to Kelvin?

```python
def fahr_to_kelvin(temp_f):
    temp_c = fahr_to_celsius(temp_f)
    temp_k = celsius_to_kelvin(temp_c)
    return temp_k

fahr_to_kelvin(99)
```

We just **composed** a function

```python
print(temp_k) 
```

`temp_k` is a **local variable**

```python
temp_kelvin = fahr_to_kelvin(212.0)
print(temp_kelvin)
```

`temp_kelvin` is a **global variable**

```python
# this function calculates kelvin degrees from fahrenheit values
def fahr_to_kelvin(temp_f):
    # fahr_to_celsius was defined elsewhere
    temp_c = fahr_to_celsius(temp_f)
    # celsius_to_kelvin was also defined elsewhere
    temp_k = celsius_to_kelvin(temp_c)
    return temp_k

fahr_to_kelvin(25)
```

now let's build a **help message** for our function 

```python
# this function calculates kelvin degrees from fahrenheit values
def fahr_to_kelvin(temp_f):
    """This function returns temperature in Kelvin from Fahrenheit
        temp_f is the argument taking the value
        if it had something else would go in here
    """
    # fahr_to_celsius was defined elsewhere
    temp_c = fahr_to_celsius(temp_f)
    # celsius_to_kelvin was also defined elsewhere
    temp_k = celsius_to_kelvin(temp_c)
    return temp_k

fahr_to_kelvin(25)
help(fahr_to_kelvin)
```

```python
def kelvin_to_celsius(temp_k):
    '''This function should reture temp in Celsius from Kelvin units
    '''
    return(temp_k - 273.15)

print(kelvin_to_celsius(-16))
```

This function is **not realistic**. Let's check with **assertions*

```python
def kelvin_to_celsius(temp_k):
    '''This function should reture temp in Celsius from Kelvin units
    '''
    assert temp_k > 0, 'Data should only contain positive values'
    return(temp_k - 273.15)

print(kelvin_to_celsius(-17))
```

We can combine it with `try` statements

```python
# This function will calculate celsius degrees from kelvin 
def kelvin_to_celsius(temp_k, assert_or_try=""):
    """ This function should return temperature in Celsius from Kelvin units; 
        temp_k is the argument taking the value
    """
    if  assert_or_try == "assert":
        assert temp_k > 0, 'Data should only contain positive values'
        return(temp_k - 273.15)
    elif assert_or_try == "try":
        try:
            print('Data should only contain positive values')
        except temp_k > 0:
            return(temp_k - 273.15)
    else:
        print("Warning: no test chosen, run at owns risk")
        return(temp_k - 273.15)
```

```python
print(kelvin_to_celsius(temp_k=16, assert_or_try=""))
```

```python
print(kelvin_to_celsius(temp_k=16, assert_or_try="try"))
```

```python
print(kelvin_to_celsius(temp_k=-16, assert_or_try="assert"))
```
