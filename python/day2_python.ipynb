```python
# analyze data  and visualize it 

import numpy as np 

## import the data ###

data=np.loadtxt(fname="data/inflammation-01.csv",delimiter=",")
print(data.shape)

## visualizing ##

import matplotlib.pyplot as plt 

plt.plot(np.mean(data,axis=0))
plt.plot(np.mean(data,axis=1))
```

```python
import glob 

# use glob to find all inflamation data files 

file_pattern="data/inflammation-*.csv"
all_files=glob.glob(file_pattern)

# initialize an empty array to store data arrays 

data_arrays=[]

# Loop through all matching files and load them as numpy arrays 

for file in all_files:
    print(file)
    data=np.loadtxt(file,delimiter=",")
    data_arrays.append(data)
    
# select samples

sample=data_arrays[:4]

for datasets in sample:
    mean_perday=np.mean(datasets,axis=0)
    max_perday=np.amax(datasets,axis=0)
    min_perday=np.amin(datasets,axis=0)
    
    ## create the figure of three subplots 
    fig=plt.figure(figsize=(15,3))
    ## plotting in subplots 
    axes1=fig.add_subplot(1,3,1)
    axes2=fig.add_subplot(1,3,2)
    axes3=fig.add_subplot(1,3,3)
    
    axes1.plot(mean_perday)
    axes2.plot(max_perday)
    axes3.plot(min_perday)
    
    
    
```

```python
## Functions 

fah_value=99
cel_value=(fah_value-32)*(5/9)

print(cel_value)

fah_values=[99,100,101]
cel_values=[]
for values in fah_values:
    cel_value=(values -32)*(5/9)
    cel_values.append(cel_value)
    
print(cel_values)
```

```python
def fah2celcius (fah_value):
    celcius_value=(fah_value -32)*(5/9)
    return celcius_value
fah2celcius(99),fah2celcius(100),fah2celcius(101)
```

```python
def celcius2kelvin (celcius_value):
    kelvin_value=celcius_value+273.15
    return kelvin_value
def fah2kelvin(fah_value):
    celcius_value=fah2celcius(fah_value)
    kelvin_value=celcius2kelvin(celcius_value)
    return kelvin_value
```

```python
print("Freezing point of water in Kelvin",fah2kelvin(32),"K")
print("Boiling point of water in Kelvin",fah2kelvin(212),"K")

kelvin_value=fah2kelvin(32)
```
