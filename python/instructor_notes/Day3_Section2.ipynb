# Creating functions

We've already learned how to use **built-in functions** (`print()`, `type()`) and **library functions** (`numpy.loadtxt()`, `matplotlib.pyplot.subplots()`). Here we'll learn how to create custom functions and how to use them

Let's start by converting temperature values

```python
fahrenheit_val = 99
celsius_val = ((fahrenheit_val - 32) * (5/9))
celsius_val
```

We can just copy the code and chage the input

```python
fahrenheit_val = 43
celsius_val = ((fahrenheit_val - 32) * (5/9))
celsius_val
```

Functions are a better way to reuse functionality other than copy-pasting

```python
def fahr_to_celsius(temp):
    return((temp - 32) * (5/9))
```

We can now call the function as may times as we want with whatever input values we'd like to

```python
print(fahr_to_celsius(99))
print(fahr_to_celsius(43))
```

We can also write the function to turn Celsius into Kelvin

```python
def celsius_to_kelvin(temp_c):
    return temp_c + 273.15

celsius_to_kelvin(27)
```

What about converting Fahrenheit to Kelvin? We could write out another formula, but we don’t need to. Instead, we can **compose** the two functions we have already created

```python
def fahr_to_kelvin(temp_f):
    temp_c = fahr_to_celsius(temp_f)
    temp_k = celsius_to_kelvin(temp_c)
    return temp_k
```

```python
fahr_to_kelvin(99)
```

This is our first taste of how larger programs are built: we define basic operations, then combine them in ever-larger chunks to get the effect we want.

Variables have a **scope** depending on where they were **defined**

```python
print('Again, temperature in Kelvin was:', temp_k)
```

`temp_k` was defined within `fahr_to_kelvin` so it is a **local** variable. If we want to reuse the temperature in Kelvin we can store the result of the **function call** in a variable:

```python
temp_kelvin = fahr_to_kelvin(212.0)
print('temperature in Kelvin was:', temp_kelvin)
```

`temp_kelvin`, being defined outside any function, is said to be **global**. Global variables can then be used within functions

```python
def print_temperatures():
    print('temperature in Fahrenheit was:', temp_fahr)
    print('temperature in Kelvin was:', temp_kelvin)

temp_fahr = 212.0
temp_kelvin = fahr_to_kelvin(temp_fahr)

print_temperatures()
```

# Documenting and testing

So far, we've been **documenting** most steps using markdown cells. We have also used **comments** within the code,

```python
# this function calculates kelvin degrees from fahrenheit values
# it can take a global variable within temp_f
def fahr_to_kelvin(temp_f):
    # fahr_to_celsius was defined elsewhere above
    temp_c = fahr_to_celsius(temp_f)
    # celsius_to_kelvin was defined elsewhere above
    temp_k = celsius_to_kelvin(temp_c)
    return(temp_k)

print('99 degrees Fahrenheit in Kelvin are: ', fahr_to_kelvin(99))
```

Now let's build a **help message** for our fahr_to_kelvin function

```python
# this function calculates kelvin degrees from fahrenheit values
# it can take a global variable within temp_f
def fahr_to_kelvin(temp_f):
    """ This function returns temperature in Kelvin from Fahrenheit units 
        temp_f is the argument taking the value
    """
    # fahr_to_celsius was defined elsewhere above
    temp_c = fahr_to_celsius(temp_f)
    # celsius_to_kelvin was defined elsewhere above
    temp_k = celsius_to_kelvin(temp_c)
    return(temp_k)

print('99 degrees Fahrenheit in Kelvin are: ', fahr_to_kelvin(99))
```

```python
help(fahr_to_kelvin)
```

We have built a **docstring** to be used by the `help()` function. This help message helps further document our `fahr_to_kelvin` function

Let's do a new, nicely documented function to calculate kevin to celsius

```python
def kelvin_to_celsius(temp_k):
    """ This function should return temperature in Celsius from Kelvin units; 
        temp_k is the argument taking the value
    """
    return(temp_k - 273.15)
```

```python
print('Kelvin to Celsius is :', kelvin_to_celsius(-16))
```

This function is not realistic though. Let's make a reality check with **assertions**

```python
def kelvin_to_celsius(temp_k):
    """ This function should return temperature in Celsius from Kelvin units; 
        temp_k is the argument taking the value
    """
    assert temp_k > 0, 'Data should only contain positive values'
    return(temp_k - 273.15)
```

```python
print('Kelvin to Celsius', kelvin_to_celsius(-16))
```

We can also use `try` statements

```python
# This function will calculate celsius degrees from kelvin 
def kelvin_to_celsius(temp_k, assert_or_try=""):
    """ This function should return temperature in Celsius from Kelvin units; 
        temp_k is the argument taking the value
    """
    if  assert_or_try == "assert":
        assert temp_k > 0, 'Data should only contain positive values'
        return(temp_k - 273.15)
    elif assert_or_try == "try":
        try:
            print('Data should only contain positive values')
        except temp_k > 0:
            return(temp_k - 273.15)
    else:
        print("Warning: no test chosen, run at owns risk")
```

```python
print('Kelvin to Celsius', kelvin_to_celsius(temp_k=-16, assert_or_try="assert"))
```

```python
print('Kelvin to Celsius', kelvin_to_celsius(temp_k=-16, assert_or_try="try"))
```
