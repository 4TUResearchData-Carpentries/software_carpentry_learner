# Python fundamentals

## Variables

Any python interpreter can be used as a calculator

```python
3 + 5 * 4
```

Operations are more fun when done with **variables**, let's assign one

```python
weight_kg = 60
```

From now on, whenever we use `weight_kg`, Python will substitute the value we assigned to it. A *variable* is a *name* for a *value*.

In Python, variable names:

- can include letters, digits, and underscores
- cannot start with a digit
- are case sensitive.

```python
weight_kg
```

# Data types

Python knows various types of data. Three common ones are:

- numbers
    - integers 
    - floating point numbers
- strings

```python
# this is an integer
weight_kg = 60
type(weight_kg)
```

```python
# this is a float
weight_kg = 60.3
type(weight_kg)
```

```python
# this is a string
weight_kg = 'sixty'
type(weight_kg)
```

```python
# this is also a string
weight_kg = "60"
type(weight_kg)
```

## Using variables

We can make calculations using data stored in variables

```python
weight_kg = 60
weight_lb = 2.2 * weight_kg
print(weight_lb)
```

We'll use variables to represent patient weight, let's create an identifier

```python
patient_id = '001'
print(patient_id)
# let's indicate wich type of patient we're talking about
patient_id = 'inflam_' + patient_id
print(patient_id)
type(patient_id)
```

## Using built-in functions

We already used `type` as a built-in function, also `print`

```python
print(weight_kg)
print(weight_lb)
print(patient_id)
```

We can combine the output of variables within `print` function calls for more readable messages

```python
print('Patient: ', patient_id, ' weighs ', weight_kg, ' kg')
```
