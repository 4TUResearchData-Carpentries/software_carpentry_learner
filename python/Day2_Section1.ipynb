# Python fundamentals I

## Variables

Any python interpreter can be used as a calculator

```python
3 + 5 * 4 
```

Operations are more fun when done with **variables**, let's assign one

```python
weight_kg = 60
```

We assigned the value **60** to the variable **weight_kg**

From now on, whenever we use `weight_kg`, Python will substitute the value we assigned to it. A *variable* is a *name* for a *value*.

In Python, variable names:

- can include letters, digits, and underscores
- cannot start with a digit
- are case sensitive.

```python
weight_kg
```

## Data types

Python knows various types of data. Three common ones are:

- numbers
    - integers 
    - floating point numbers
- strings

```python
# this is an integer
weight_kg = 60
type(weight_kg)
```

```python
# this is a float
weight_kg = 60.3
type(weight_kg)
```

```python
# this is a string
weight_kg = 'sixty'
type(weight_kg)
```

```python
# this is also a string. Careful
weight_kg = "60"
type(weight_kg)
```

## Using variables

We can make calculations using data stored in variables

```python
weight_kg = 60
weight_lb = 2.2 * weight_kg
```

We'll use variables to represent patient weight, let's create a patient identifier

```python
patient_id = '001'
# let's indicate which type of patient we're referring to 
patient_id = 'inflam_' + patient_id
```

## Using built-in functions

We already used the built-on function `type`. There's many more, for example, `print`

```python
print(weight_kg)
print(weight_lb)
print(patient_id)
```

We can combine multiple variables within `print` calls for more readable messages

```python
print('Patient: ', patient_id, ' weighs ', weight_kg, ' kg')
```

We can do arithmetics

```python
print('weight in pounds is:', 2.2 * weight_kg)
```

We can also call functions within functions

```python
print(type(weight_kg))
print(type(patient_id))
```
